```python
# 类 Class

# 作用域和命名空间示例

def scope_test():
    def do_local():
        spam = "local spam"

    def do_nonlocal():
        nonlocal spam
        spam = "nonlocal spam"

    def do_global():
        global spam
        spam = "global spam"
    spam = "test spam"
    do_local()
    print("After local assignment:", spam)
    do_nonlocal()
    print("After nonlocal assignment:", spam)
    do_global()
    print("After global assignment:", spam)
    
scope_test()
print("In global scope:", spam)
```

```python
# 初探类

# class ClassName:
#     <statment-1>
#     ...
#     <statment-n>

# 类对象支持两种操作：属性引用和实例化。

# 属性引用 使用 Python 中所有属性引用所使用的标准语法: obj.name。
# 有效的属性名称是类对象被创建时存在于类命名空间中的所有名称。 因此，
# 如果类定义是这样的:

class MyClass:
    """A simple example class"""
    i = 12345

    def f(self):
        return "Hello, World"

# 那么 MyClass.i 和 MyClass.f 就是有效的属性引用，将分别返回一个整数和
# 一个函数对象。 类属性也可以被赋值，因此可以通过赋值来更改 MyClass.i 的
# 值。 __doc__ 也是一个有效的属性，将返回所属类的文档字符串: 
# "A simple example class"。

mc = MyClass()
print(type(mc), mc)
print(type(mc.i), mc.i)
print(type(mc.f), mc.f)
print(type(mc.__doc__), mc.__doc__)
print(type(mc.__dict__), mc.__dict__)
```

```python
# 实例化操作 (“调用”类对象) 会创建一个空对象。 许多类都希望创建的对象实例是根据特
# 定初始状态定制的。 因此一个类可能会定义名为 __init__() 的特殊方法，就像这样:

# java 里的类的构造器呗，牛逼啥呀

class MyClass:
    """A simple example class"""
    def __init__(self):
        self.data = []
        
    def f(self):
        return "Hello, World"
x = MyClass()
print(type(x), x)
print(type(x.data), x.data)
print(type(x.__dict__), x.__dict__)
```

```python
# 当然，__init__() 方法还有一些参数用于实现更高的灵活性。 在这种情况下，提供给类实例
# 化运算符的参数将被传递给 __init__()。 例如，

class Complex:
    def __init__(self, realpart, imagpart):
        self.r = realpart
        self.i = imagpart

x = Complex(3.0, 4.5)
print(x.r, x.i)

l = [1, -1]
y = Complex(*l)
print(y.r, y.i)

d = {"realpart": 6, "imagpart": -9}
z = Complex(**d)
print(z.r, z.i)
```

```python
# 实例对象所能理解的唯一操作是属性引用。 有两种有效的属性名称：数据属性和方法。

# 方法对象

# 在 MyClass 示例中，这将返回字符串 'hello world'。 但是，方法并不是必须立即
# 调用: x.f 是一个方法对象，它可以被保存起来以后再调用。 例如:

x = MyClass()

xf = x.f
for i in range(3):
    print(xf())
```

```python
# 正如 名称和对象 中已讨论过的，共享数据可能在涉及 mutable 对象例如列表和字典的时候导致
# 令人惊讶的结果。 例如以下代码中的 tricks 列表不应该被用作类变量，因为所有的 Dog 实例
# 将只共享一个单独的列表:

class Dog:

    tricks = []             # mistaken use of a class variable

    def __init__(self, name):
        self.name = name

    def add_trick(self, trick):
        self.tricks.append(trick)

>>> d = Dog('Fido')
>>> e = Dog('Buddy')
>>> d.add_trick('roll over')
>>> e.add_trick('play dead')
>>> d.tricks                # unexpectedly shared by all dogs
['roll over', 'play dead']
```

```python
# 继承 inheritance

# 当然，如果不支持继承，语言特性就不值得称为“类”。派生类定义的语法如下所示:
# ??? 你拿 GO 当什么了？

# class DerivedClassName(BaseClassName):
#     <statement-1>
#     .
#     .
#     .
#     <statement-N>

# class DerivedClassName(modname.BaseClassName):

# 使用 isinstance() 来检查一个实例的类型: isinstance(obj, int) 仅会在 obj.__class__ 、
# 为 int 或某个派生自 int 的类时为 True。

# 使用 issubclass() 来检查类的继承关系: issubclass(bool, int) 为 True，因为 bool 是 int
# 的子类。 但是，issubclass(float, int) 为 False，因为 float 不是 int 的子类。

class A():
    pass

class B(A):
    pass

a = A()
b = B()

print(isinstance(a, A))
print(isinstance(a, B))

print(isinstance(b, A))
print(isinstance(b, B))

print(issubclass(B, A))
print(issubclass(A, B))
```

```python
# 多重继承

# class DeivedClassName(Base1, Base2, Base3):
#     <statment-1>
#     .
#     .
#     .
#     <statment-N>

# 对于多数目的来说，在最简单的情况下，你可以认为搜索从父类所继承属性的操作是深度优先、
# 从左到右的，当层次结构存在重叠时不会在同一个类中搜索两次。 因此，如果某个属性在 De
# rivedClassName 中找不到，就会在 Base1 中搜索它，然后（递归地）在 Base1 的基类
# 中搜索，如果在那里也找不到，就将在 Base2 中搜索，依此类推。

# DFS 搜索呗

# 那种仅限从一个对象内部访问的“私有”实例变量在 Python 中并不存在。 但是，大多数 Python
# 代码都遵循这样一个约定：带有一个下划线的名称 (例如 _spam) 应该被当作是 API 的非公有部
# 分 (无论它是函数、方法或是数据成员)。 这应当被视为一个实现细节，可能不经通知即加以改变。
```

```python
# __xxx 是什么？
# Python 提供了有限的私有机制，双下划线开头

# self.__money

class Person:
    def __init__(self):
        self.__money = 100

p = Person()
# print(p.__money)

# 看起来像私有。
# 但是实际上：
# 它不是消失了。
# Python 偷偷改名字了。

# name mangling

# 编译的时候
# class Person:
#     def __init__(self):
#         self.__money = 100

# 自动变成 self._Person__money = 100
# self._ClassName__fieldname

print(p.__dict__)
# 可以访问
print(p._Person__money)
```

```python
# 为什么需要这个东西？

# 假设：

class Mapping:
    """"""
    def __init__(self, iterable):
        self.items_list = []

        self.__update(iterable)

        def update(self, iterable):
            # for item in iterable:
                self.items_list.append(item)
        __update = update

# 这里：
# __update
# 是什么意思呢？保存一个一个原始版本，等价于：
# self.__update = self.update
# 但是它被名字改写：
# 实际上：
# _Mapping__update

# 然后子类
class MappingSubclass(Mapping):

    def update(self, keys, values):
        for item in zip(keys, values):
            self.items_list.append(item)

# 子类重写了update()

# 子类想要调用父类的方法
# self.__update() 变成 self._Mapping__update()

# 子类 __update
# 会变 _MappingSubclass__update

# 改写规则
# 任何形式为 __spam

# 两个前导下划线
# 最多一个结尾下划线

# __x
# __name
# __value_

# _ClassName__x
# _ClassName__name
# _ClassName__value_
```

```python
# 实例对象方法是什么？

class A:
    def hello(self):
        print("hello")

# 类 A.hello 是什么
# 它是一个函数
print(A.hello)

# 但是
a = A()
# a.hello 不一样, 叫实例方法对象
print(a.hello)

# 因为 Python 自动帮你绑定了：self
# 原本
A.hello(a)
# 需要自己传对象
# 但是
a.hello() # python 自动变
# A.hello(a)

# 所以 a.hello 内部其实保存两个东西
# m.__self__ 就是这个方法绑定到哪个*对象*
# m, stands for method
# print(A.hello.__self__) # 会报错
print(a.hello.__self__)

# m.__func__ 就是原来的函数

print(a.hello.__func__)

a.hello() # 实际上就是
a.hello.__func__(a.hello.__self__)
```

```python
# Python 里面非常核心的一块：
# 迭代器（iterator）→ 生成器（generator）→ 生成器表达式（generator expression）。

# Python 的 for 循环为什么可以遍历一切东西？

# for 背后的秘密

# 平时
for x in [1,2,3]:
    print(x)

# 看起来很简单，但是Python实际干的是

it = iter([1, 2, 3])

while True:
    try:
        x = next(it)
        print(x)
    except StopIteration:
        break
```

```python
# 自己写一个迭代器

class Reverse:
    def __init__(self, data):
        self.data=data
        self.index=len(data)

    def __iter__(self):
        return self # 意思是我是我的迭代器， iter(rev)返回自己，
        # iter()函数返回一个迭代器，就是调用对象内部的__iter__方法

    def __next__(self): # 关键
        if self.index == 0:
            raise StopIteration
        self.index -= 1
        return self.data[self.index]

rev = Reverse("spam")
for i in rev:
    print(i)
```

```python
# 生成器 generator 是什么？

# 上面自己写 iterator：
# 麻烦。
# 因为你必须：
# 保存状态：

# self.index
# self.data

# Python提供
# yield, 自动帮你保存，返回一个generator 对象

# 普通函数

# def f():
#     return 1
#     return 2 # 只能返回一次

# 但是，生成器
def f():
    yield 1
    yield 2 # 可以暂停

g=f() # 没有执行，只是得到generator对象

# next() 函数通过调用 iterator 的 __next__() 
# 方法获取下一个元素。如果迭代器耗尽，则返回给定的 
# default，如果没有默认值则触发 StopIteration。

print(next(g))
print(next(g))
print(next(g))
```
