```python
# 类其实就是一个 namespace

class Student:
    school = "CAS"

    def hello(self):
        print("hi")

# 这个类内部其实就是一个字典：

print(Student.__dict__)
# 所以
print(Student.school)
# 实际上
print(Student.__dict__["school"])
```

```python
# 属性查找机制（非常重要）

# obj.x
# 不是直接找
# 顺序

# obj.__dict__
#         |
#         ↓
# obj的class
#         |
#         ↓
# 父类
#         |
#         ↓
# __getattr__

class A:
    x = 100

a = A()

a.x
# 对象 a.__dict__ 为空
# 找 A.__dict__, 找到 x = 100

>>> a.x # 返回100
```

```python
# 方法为什么自动带 self？

class A:
    def hello(self):
        print("hello")

# 类里面 A.hello 只是函数

# 但是
a = A()
a.hello # 变成 bound method

# __self__
#      |
#      a

# __func__
#      |
#      hello函数

# 调用
a.hello()
# 等价
A.hello(a)
# 等价
a.hello.__func__(a.hello.__self__)
```

```python
# 魔术方法（dunder methods）

# 之前见过

# __init__
# __iter__
# __next__

# 它们不是给你直接调用的。而是让 Python 语法调用你。

# len(a)
# Python 调用 a.__len__()

class MyList():

    def __len__(self):
        return 666

x = MyList()
print(len(x))

# 加法

# a + b
# 实际上是 a.__add__(b)

class Vector:

    def __init__(self, x):
        self.x = x

    def __add__(self, other):
        return self.x + other.x

a = Vector(6)
b = Vector(8)

print(a + b)

# 下标

# obj[5] 
# 实际上是 obj.__getitem__(5)

class Test:

    def __getitem__(self, index):
        return index * 2

t = Test()
print(t[6])
print(t[24])
```

```python
# new 和 init 的区别

# 创建对象
# =
# __init__ 

# 我不这么以为，我学过 Java

# 实际

# 调用类
#  |
#  ↓
# __new__
# 创建对象
#  |
#  ↓
# __init__
# 初始化对象

class A:

    def __new__(cls):
        print("new")
        return super().__new__(cls)

    def __init__(self):
        print("init")

a = A()

# 为什么需要 __new__？

# 例如，单例模式

class Singleton:

    instance=None

    def __new__(cls):
        if cls.instance is None:
            cls.instance = super().__new__(cls)

        return cls.instance

# 无论创建多少次
# 都是 a is b
```

```python
# property 非常重要

# python不喜欢 get_x() set_x() 这种Java风格

class Person:
    """"""
    def __init__(self):
        self._age = 0

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        self._age=value

p = Person()
p.age = 20
print(p.age)
```

```python
# classmethod 和 staticmethod

# 普通方法
# def foo(self)
# 绑定对象 
# obj.foo()

# classmethod

# @classmethod
# def foo(cls):

# 绑定类：
# Class.foo()

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    @classmethod
    def from_string(cls, s):
        name, age = s.split()
        return cls(name, age)

p = Person.from_string("ash 24")
print(p.__dict__)
# 类似 Java 静态工厂

# staticmethod
# 没有self
# @staticmethod
# def add(a, b):
    # return a + b

# 纯工具函数
```
