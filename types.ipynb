```python
# list is a reference

rgb = ['red', 'green', 'blue']
rgba = rgb
id(rgb) == id(rgb)
```

```python
rgba.append("Alph")
rgb
```

```python
# 切片操作返回包含请求元素的新列表。以下切片操作会返回列表的 浅拷贝
correct_rgba = rgba[:]
correct_rgba[-1] = "alpha"
correct_rgba
```

```python
# 为切片赋值可以改变列表大小，甚至清空整个列表：
letters = ['a', 'b', 'c', 'd', 'e', 'f', 'g']
print(letters)

# 替换一些值
letters[2:5] = ['C', 'D', 'E']
print(letters)

letters[2:5] = ['F', 'F', 'F', 'F']
print(letters)

# 现在移除它们
letters[2:5] = []
print(letters)

# 通过用一个空列表替代所有元素来清空列表
letters[:] = []
print(letters)
```
