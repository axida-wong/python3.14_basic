```python
# 定义函数

# generates fibonacci numbers

# 函数内的第一条语句就是文档字符串，docstring. 利用文档字符串可以自动盛恒在线文档或者打印版文档。
# 实参是按照值调用来传递的，其中的的值始终是对象object的引用，而不是对象object的值

def fib(n):
    """Print a Fibonacci series less than n"""
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a + b
    print()

fib(2000)
```

```python
def fib2(n):
    """Retrun a list containing the Feibnacci series up to n."""
    result = []
    a, b = 0, 1
    while a < n:
        result.append(a)
        a, b = b, a + b

    return result

f100 = fib2(100)
f100

# 新特性： return 语句没有表达式参数时， 返回None, 函数执行完退出也返回None
```

```python
# 函数定义详解

# 1. 默认值参数

def ask_ok(prompt, retries=4, reminder="Please try again!"):
    while True:
        reply = input(prompt)
        if reply in {'y', 'ye', 'yes'}:
            return True
        if reply in {'n', 'no', 'nop', 'nope'}:
            return False
        retries -= 1
        if retries < 0:
            raise ValueError('invalid uesr response')
        print(reminder)
```

```python
# 默认值只计算一次， 默认值为列表，字典或类实例等可变对象时， 会产生与该规则不同的结果

def f(a, L=[]):
    L.append(a)
    return L

print(f(1))
print(f(2))
print(f(3))
```

```python
# 如果后续调用不想共享默认值， 应这样做

def f(a, L=None):
    if L is None:
        L = []

    L.append(a)

    return L

print(f(1))
print(f(2))
print(f(3))
```

```python
# 2. 关键字参数
# kwarg=value 形式的参数

def parrot(voltage, state='a stiff', action='voom', type='Norwegin Blue'):
    print("-- This parrot wouldn't", action, end=' ')
    print("if you put", voltage, "volts through it.")
    print("-- Lovely plumage, the", type)
    print("-- It's", state, "!")
```

```python
# 当存在形式为 **names 的最后一个形参时， 它讲接受一个字典（就是之前的 *rest, **rest, 
# 收集与展开的语法）。它可以接受包含正式形参列表外的关键字参数。

# 它还可以接受包含正式形参列表以外的位置参数的 tuple 的 *name 形式的正式形参组合使用，
# *name 必须出现在 **name之前。

def cheeseshop(kind, *argumenets, **keywords):
    print("-- Do you have any", kind, "?")
    print("-- I'm sorry, we're all out of", kind)
    for arg in argumenets:
        print(arg)
    print("*" * 40)
    for kw in keywords: # 啥也不动，迭代keys, dict.items()迭代值
        print(kw, ":", keywords[kw])
```

```python
cheeseshop("Limburger", "It's very runny, sir.",
           "It's really very, VERY runny, sir.",
           shopekeeper="Michael Palin",
           client="John Cleese",
           sketch="Cheese Shop Sketch")
```

```python
# 特殊参数, 限制参数的传递方式
# 仅限位置参数, 在最后一个仅限关键字参数前后 /
# 仅限关键字参数, 在第一个仅限关键字参数前加 *

def f(pos1, pos2, /, pos_or_kwd, *, kwd1, kwd2):
    """|仅限位置|     |位置或关键字|   |仅限关键字|"""
    print(pos1, end=', ')
    print(pos2, end=', ')
    print(pos_or_kwd, end=', ')
    print(kwd1, end=', ')
    print(kwd2, end=', ')
    print()

def standerd_arg(arg):
    print(arg)

def pos_only_arg(arg, /):
    print(arg)

def kwd_only_arg(*, arg):
    print(arg)

def combined_exanple(pos_only, /, standerd, *, kwd_only):
    print(pos_only, standerd, kwd_only)

f('one', 'two', 'three', kwd1='four', kwd2='five')
f('one', 'two', pos_or_kwd='three', kwd1='four', kwd2='five')
print()
print("*" * (20 - len("standerd") // 2), "standerd", "*" * (20 - len("standerd") // 2))
standerd_arg(2)
standerd_arg(arg=2)
print()
print("*" * (20 - len("pos_only") // 2), "pos_only", "*" * (20 - len("pos_only") // 2))
pos_only_arg(2)
# pos_only_arg(arg=2)
print()
print("*" * (20 - len("kwd_only") // 2), "kwd_only", "*" * (20 - len("kwd_only") // 2))
# kwd_only_arg(3)
kwd_only_arg(arg=3)
print()
print("*" * (20 - len("combined") // 2), "combined", "*" * (20 - len("combined") // 2))
combined_exanple(2, 3, kwd_only=4)
combined_exanple(2, standerd=3, kwd_only=4)


```

```python
# 下列定义中，kwds把name当作键， 可能和位置参数name有冲突

def foo(name, **kwds):
    return 'name' in kwds

# 函数永远返回False, 因为关键字'name'总与抵押给形参绑定
# 加上 / （仅限位置参数）就可以了。此时函数定义把当作位置
# 参数，'name' 也可以作为关键词参数的键

def foo(name, /, **kwds):
    return 'name' in kwds

foo(1, **{'name': 'ash wong'})
```

```python
# 任意实参列表

# 调用函数时，使用任意数量实参是最少见的选项。这些实参包含在元组中。
# 在可变数量的实参之前，可能有若干个普通参数

def write_multiple_items(file, separator, *args):
    file.write(separator.join(args)) #
    # Python join() 方法用于将序列中的元素以指定的字符连接生成一个新的字符串。

# variadic 参数用于采集官邸给函数的左右剩余参数，因此，他们通常放在形参列表的末尾
# *args 之后的任何形式参数仅限关键字参数

def concat(*args, sep='/'):
    return sep.join(args)

l=['', 'usr', 'bin', 'bash']
print(concat("earth", "mars", "venus"))
print(concat("earth", "mars", "venus", sep='.'))
print(*l, sep='_'), #有解包操作
print(concat(*l))
```

```python
# Lambda 表达式

# lambda 关键字用于创建小巧的匿名函数，lambda: a, b: a + b, 返回两个参数的和
# lambda 表达式可以用在任何，需要函数对象的地方，和函数嵌套定义一样，lambda可以
# 引用包含作用域中的变量

def make_incrementor(n):
    """注意哈，这个表达式返回一个lambda表达式"""
    return lambda x: x + n

f = make_incrementor # 传递函数指针后
print(f(2), f(3), sep='\n') # f(2) 就是一个lamda表达式 --> lamda x: x + 2
print()

f = make_incrementor(42)
print(f(2), f(3), sep='\n')

# 上例 lambda 表达式返回一个函数，另一个用法就是, 传入一个小函数作为参数，
# 回调函数呗---老C/C++
# list.sort() 就接受一个排序函数key, 它可以是一个lambda函数

pairs = [(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four'), (5, 'five')]
pairs.sort()
print(pairs)
pairs.sort(key=lambda pair: pair[1])
print(pairs)

l = [34, 54, 67, 98, 23, 11, 39, 34, 66, 78]
l.sort()
print(l)
l.sort(key=lambda a: -a)
print(l)
```

```python
# 以下是文档字符串内容和格式的约定。

# 第一行应为对象用途的简短摘要。为保持简洁，不要在这里显式说明对象名或类型，因为
# 可通过其他方式获取这些信息（除非该名称碰巧是描述函数操作的动词）。这一行应以大
# 写字母开头，以句点结尾。

# 文档字符串为多行时，第二行应为空白行，在视觉上将摘要与其余描述分开。后面的行可
# 包含若干段落，描述对象的调用约定、副作用等。对于被用作模块、类或函数文档字符串
# 的多行字符串字面值 Python 解析器会去除其中的缩进。

def my_function():
    """Do nothing, but document it.

    No, really, it doesn't do anything:

        >>> my_function()
        >>>
    """
    pass

print(my_function.__doc__) # 真就一切皆对象了呗，不过java更好吧，
# Everything belongs to class


```

```python
# 函数注解 
# 
# 是可选的用户自定义函数类型的元数据完整信息

# 标注 是以字典形式存放在函数的 __annotations__ 属性中的而对函数的其他部分
# 没有影响。 
# 
# 形参标注的定义方式是在形参名后加冒号，后面再跟一个将被求值为标注的值的表达式。
# 返回值标注的定义方式是加符号 ->，后面再跟一个表达式，它位于形参列表和表示def
# 语句结束的冒号之间。 

def f(ham: str, eggs: str = 'eggs') -> str:
    print("Annotations: ", f.__annotations__)
    print("Arguments: ", ham ,eggs)
    return ham + ' and ' + eggs

f('spam')
```
